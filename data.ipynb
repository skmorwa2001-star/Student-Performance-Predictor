```python
features=['G1','G2','health','freetime','failures','studytime','traveltime','absences','goout']
```

```python
# Select the features

X=df[features]
```

```python
# Target column
Y=df['G3']
```

```python
### models 
```

```python
from sklearn.model_selection import train_test_split


X_train,X_test,Y_train,Y_test= train_test_split(X,Y,test_size=0.2)
```

```python
X.dtypes
```

```python
# Linear Regression

from sklearn.linear_model import LinearRegression

model_lr=LinearRegression()

model_lr.fit(X_train,Y_train)
```

```python
y_pred_lr=model_lr.predict(X_test)
```

```python
# Accuracy Score

from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

mae_lr=mean_absolute_error(Y_test,y_pred_lr)
mse_lr=mean_squared_error(Y_test,y_pred_lr)
rmse_lr=root_mean_squared_error(Y_test,y_pred_lr)
r2_lr=r2_score(Y_test,y_pred_lr)

print("Linear Regression Performance")
print("MAE:", mae_lr)
print("MSE:", mse_lr)
print("RMSE:",rmse_lr)
print("R2 Score:",r2_lr)
```

```python
# Decision Tree Regressor

from sklearn.tree import DecisionTreeRegressor

model_dt=DecisionTreeRegressor(max_depth=5, random_state=42)

model_dt.fit(X_train, Y_train)

y_pred_dt=model_dt.predict(X_test)
```

```python
# Accuracy Score for Decision Tree regressor

mae_dt=mean_absolute_error(Y_test,y_pred_dt)
mse_dt=mean_squared_error(Y_test,y_pred_dt)
rmse_dt=root_mean_squared_error(Y_test,y_pred_dt)
r2_dt=r2_score(Y_test,y_pred_dt)

print("Decision Tree Regreesor:")
print("MAE:", mae_dt)
print("MSE:", mse_dt)
print("RMSE:",rmse_dt)
print("R2 Score:",r2_dt)
```

```python
# Random Forest Regressor

from sklearn.ensemble import RandomForestRegressor

model_rf=RandomForestRegressor(n_estimators=200, random_state=42, max_depth=5)

model_rf.fit(X_train, Y_train)

y_pred_rf=model_rf.predict(X_test)
```

```python
# Accuracy Score for Random Forest regressor

mae_rf=mean_absolute_error(Y_test,y_pred_rf)
mse_rf=mean_squared_error(Y_test,y_pred_rf)
rmse_rf=root_mean_squared_error(Y_test,y_pred_rf)
r2_rf=r2_score(Y_test,y_pred_rf)

print("Random Forest Regreesor:")
print("MAE:", mae_rf)
print("MSE:", mse_rf)
print("RMSE:",rmse_rf)
print("R2 Score:",r2_rf)
```
