```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
```

```python
data=pd.read_csv('student_performance_updated_1000.csv')
```

```python
df=pd.DataFrame(data)
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(['StudentID','Name'],axis=1,inplace=True)
```

```python
df[['StudyHoursPerWeek','Study Hours']].corr()
```

```python
df[['AttendanceRate','Attendance (%)']].corr()
```

```python
# Numerical misssing value

num_cols=df.select_dtypes(include='number').columns

df[num_cols]=df[num_cols].fillna(df[num_cols].median())
```

```python
# Categorical missing value

cat_cols=df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['FinalGrade'].value_counts()
```

```python
# Select the features

X=df.drop(['FinalGrade'],axis=1)
```

```python
# Apply one-hot encoding
X=pd.get_dummies(X,drop_first=True)
```
