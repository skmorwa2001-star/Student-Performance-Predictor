```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
```

```python
data=pd.read_csv('student_performance_updated_1000.csv')
```

```python
df=pd.DataFrame(data)
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Numerical misssing value

num_cols=df.select_dtypes(include='number').columns

df[num_cols]=df[num_cols].fillna(df[num_cols].median())
```

```python
# Categorical missing value

cat_cols=df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Remove unnecessary columns

df=df.drop(['StudentID','Name'],axis=1)
```

```python
df.head()
```

```python
# Check duplicate

df[['StudyHoursPerWeek','Study Hours']].corr()
```

```python
df[['AttendanceRate','Attendance (%)']].corr()
```

```python
# Select the features

X=df.drop(['FinalGrade'],axis=1)
```

```python
# Apply one-hot encoding
X=pd.get_dummies(X,drop_first=True)
```

```python
# Target column
Y=df['FinalGrade']
```

```python
X
```

```python
Y
```

```python
### models 
```

```python
from sklearn.model_selection import train_test_split


X_train,X_test,Y_train,Y_test= train_test_split(X,Y,test_size=0.2)
```

```python
X.dtypes
```

```python
# Linear Regression

from sklearn.linear_model import LinearRegression

model_lr=LinearRegression()

model_lr.fit(X_train,Y_train)
```

```python
y_pred_lr=model_lr.predict(X_test)
```

```python
# Accuracy Score

from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

mae_lr=mean_absolute_error(Y_test,y_pred_lr)
mse_lr=mean_squared_error(Y_test,y_pred_lr)
rmse_lr=root_mean_squared_error(Y_test,y_pred_lr)
r2_lr=r2_score(Y_test,y_pred_lr)

print("Accuracy Score for Linear Regression")
print("MAE:", mae_lr)
print("MSE:", mse_lr)
print("RMSE:",rmse_lr)
print("R2 Score:",r2_lr)
```
